```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
print("\n--- Missing Values (True/False) ---")
print(df.isnull())

print("\n--- Number of Missing Values in Each Column ---")
print(df.isnull().sum())
```

```python
mean_age = df['Age'].mean()

print("\nMean Age:", mean_age)

# Replace missing Age values with mean Age
df['Age'] = df['Age'].fillna(mean_age)

print("\nAfter Mean Imputation of Age:")
print(df)
```

```python
median_salary = df['Salary'].median()

print("\nMedian Salary:", median_salary)

# Replace missing Salary values with median Salary
df['Salary'] = df['Salary'].fillna(median_salary)

print("\nAfter Median Imputation of Salary:")
print(df)
```

```python
data_ffill = {
    'Student': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena'],
    'Department': ['CSE', 'ECE', None, 'ISE', None]
}

df_ffill = pd.DataFrame(data_ffill)

print("\n--- Before Forward Fill ---")
print(df_ffill)

# Forward fill missing Department values
df_ffill['Department'] = df_ffill['Department'].ffill()

print("\n--- After Forward Fill ---")
print(df_ffill)
```

```python
data_encoding = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}

df_label = pd.DataFrame(data_encoding)

print("\n--- Original Data for Label Encoding ---")
print(df_label)

# Create LabelEncoder
encoder = LabelEncoder()

# Convert Department categories into numbers
df_label['Department_Encoded'] = encoder.fit_transform(
    df_label['Department']
)

print("\n--- After Label Encoding ---")
print(df_label)

# Display the mapping/classes
print("\nClasses:")
print(encoder.classes_)
```
